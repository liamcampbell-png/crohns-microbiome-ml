**Crohn's Disease** - type of inflammatory bowel disease where the immune system attacks the digestive tract, causing chronic inflammation, pain, and digestive issues. It can affect any part of the gut but often hits the small intestine and colon. 

**Microbiome** - The full community of bacteria, viruses, fungi, and other microbes living in/on a body (in this project, specifically the gut).

**HMP2** - The second phase of the Human Microbiome Project, a large NH-funded effort that tracked microbiome changes over time in people with IBD and other conditions. It's the specific dataset your project is built on. 

**iHMP** - "Integrative" HMP - the official name for the HMP2 phase, emphasizing that it combines multiple data types (microbiome, genetics, immune markers, etc) per patient rather than just one. 

**IBD** - Inflammatory Bowel Disease, an umbrella term covering Crohn's Disease and Ulcerative Colitis. Both involve chronic gut inflammation but differ in where and how it presents. 

**OTU (Operational Taxonomic Unit) Abundance Table** - a table where rows are microbial "species" and columns are samples, with values showing how much of each microbe was found. Basically who is in the gut and how many. 

**Metadata** - the non-microbiome info attached to each sample: patient ID, diagnosis (CD/UC/Healthy), timepoint, age, etc. It's what lets you connect a microbiome profile back to a real clinical label. 

**Data leakage** - When information from outside the training set (or that wouldn't be available at real prediction time) accidentally "leaks" into training, making a model look better than it really is. For this project, it's the same patient's samples ending up in both train and test.

**AUC (Area Under the Curve)** - A single number (0 to 1) summarizing how well a model separates two classes (CD v Healthy for this project) across all decision thresholds. 0.5 is a random guessing, 1.0 is perfect separation.

**Feature Importance** - A score for each input variable (each taxon) showing how much it influenced the model's predictions. Higher importance means the model relied on that feature more to make its decisions. 

**Taxon (Plural: Taxa)** - a single named group in a biological classificationn, like one species or genus of bacteria. 

**Naive Model** - the version trained with the flawed, leaky data split (random sample-level split) - the one that looks deceptively good. 

**Honest Model** - the version trained with the corrected, leakage-free split (grouped by patient) - the one that shows real-world-like performance. 

**Feature Importance Stability** - Whether the ranking of which features matter most stays consistent across different conditions (naive v honest split). Low stability means the model's story about what drives predictions changes depending on how you split the data.
